```python
# Creamos un mini-diccionario que guarda el método SIN paréntesis (aún no se ejecuta).
directorio = {"pesados": detective.buscar_mas_pesados}

# Simulamos que la IA nos devolvió solo un texto con el nombre de la herramienta.
texto_de_la_ia = "pesados"

# Traducimos ese texto buscando la función real dentro del diccionario.
funcion_encontrada = directorio.get(texto_de_la_ia)

# Imprimimos qué es 'funcion_encontrada' (verás que Python dice que es un <bound method>).
print("Lo que sacamos del diccionario:", funcion_encontrada)

# Ahora sí la ejecutamos pasándole argumentos y vemos el resultado real.
print(funcion_encontrada(limite=2))
```

```python
# Importamos nuestras 3 clases modulares para ver el flujo completo de roles en vivo.
from detective_pc.configuracion import Configuracion

# Importamos la caja de herramientas de archivos.
from detective_pc.herramientas import DetectiveArchivos

# Importamos la clase del agente detective.
from detective_pc.agente import AgenteDetective

# Instanciamos la configuración, el detective y el agente.
config = Configuracion()
# Creamos el detective apuntando a la carpeta Descargas.
detective = DetectiveArchivos(config.carpeta_objetivo)


```

```python
# Conectamos ambos dentro del agente.
agente = AgenteDetective(config, detective)
```

```python
# Hacemos una pregunta real que obligue al agente a usar una herramienta.
print(agente.responder("¿Qué archivos PDF hay en mi carpeta?"))

```

```python
# Imprimimos uno por uno todos los mensajes acumulados en 'agente.historial' para ver cada rol.
print("\n=== GUION DE ROLES EN agente.historial ===")
# 'enumerate' nos da la posición (0, 1, 2...) y el mensaje guardado.
for indice, msg in enumerate(agente.historial):
    # Extraemos el rol dependiendo de si es un diccionario nuestro o un objeto Message devuelto por Ollama.
    rol = msg["role"] if isinstance(msg, dict) else msg.role
    # Extraemos el texto del contenido.
    contenido = msg["content"] if isinstance(msg, dict) else (msg.content or "")
    # Verificamos si el mensaje trajo una petición de herramienta ('tool_calls').
    llamadas = None if isinstance(msg, dict) else getattr(msg, "tool_calls", None)
    # Mostramos el índice, el rol, un fragmento del texto y si pidió herramienta.
    print(f"[{indice}] ROL: {rol:9} | TEXTO: {repr(contenido[:60])} | TOOLS: {llamadas}")
```

```python
# Hacemos una llamada directa a Ollama sin herramientas para ver un 'respuesta.message' normal.
r_normal = ollama.chat(model=config.modelo, messages=[{"role": "user", "content": "Hola"}], think=False)

# '.model_dump()' convierte el objeto Message de Ollama en un diccionario normal para ver sus 5 campos exactos.
print("Sin tool:", r_normal.message.model_dump())
```

```python

# Ahora hacemos una llamada con herramienta para comparar cómo cambian 'content' y 'tool_calls'.
r_con_tool = ollama.chat(model=config.modelo, messages=[{"role": "user", "content": "Lista mi carpeta"}], tools=agente.lista_tools, think=False)

# Imprimimos el diccionario interno cuando SÍ pide herramienta.
print("\nCon tool:", r_con_tool.message.model_dump())
```

```python
# Hacemos UNA pregunta doble para ver cómo Ollama puede pedir múltiples herramientas en una lista.
r_doble = ollama.chat(model=config.modelo, messages=[{"role": "user", "content": "Dime los 2 archivos más pesados y también busca qué archivos .pdf hay."}], tools=agente.lista_tools, think=False)

# Imprimimos cuántos elementos trajo la lista 'tool_calls' (el 'or []' evita error si viniera None).
print("Cantidad de herramientas en la lista:", len(r_doble.message.tool_calls or []))

# Recorremos la lista con un 'for' para ver cada petición una por una.
for i, llamada in enumerate(r_doble.message.tool_calls or []):
    # Imprimimos el número de vuelta, el nombre de la función y los argumentos que pidió.
    print(f"Vuelta {i}: {llamada.function.name} con {llamada.function.arguments}")
```

```python
# Prueba 2: Simulamos dos turnos seguidos de conversación para verificar que el agente recuerda el contexto previo.
from detective_pc.configuracion import Configuracion

# Importamos las herramientas para armar una prueba controlada de 2 turnos.
from detective_pc.herramientas import DetectiveArchivos

# Importamos la clase del agente.
from detective_pc.agente import AgenteDetective

# Creamos los 3 objetos exactamente igual que lo hace 'main.py'.
cfg = Configuracion()
# Creamos el detective con la carpeta Descargas.
det = DetectiveArchivos(cfg.carpeta_objetivo)
# Creamos una instancia limpia del agente.
ag = AgenteDetective(cfg, det)

# Turno 1: Pedimos los 3 archivos más pesados.
print("--- TURNO 1 ---")
# Ejecutamos la primera pregunta.
print(ag.responder("¿Cuáles son los 3 archivos más pesados de mi carpeta?"))

# Turno 2: Pregunta de seguimiento SIN repetir la palabra 'archivos' ni 'carpeta' para probar su memoria.
print("\n--- TURNO 2 (Prueba de memoria) ---")
# El agente debe recordar lo que respondió en el Turno 1 para contestar esto.
print(ag.responder("De esos que me mostraste, ¿cuál ocupa más espacio?"))

# Comprobamos cuántos mensajes acumuló en total 'ag.historial' tras ambos turnos.
print("\nTotal de mensajes en memoria:", len(ag.historial))
```
